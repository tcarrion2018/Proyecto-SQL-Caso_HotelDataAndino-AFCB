```python
ejercicio 17. ¿Los huéspedes con hijos cancelan más?
```

```python
%sql
SELECT
    CASE
        WHEN children > 0 OR babies > 0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END AS tipo_huesped,
    COUNT(*) AS total_reservas,
    SUM(is_canceled) AS reservas_canceladas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion_pct
FROM cursodataacademy.proyecto_hotel.hotel_bookings
GROUP BY
    CASE
        WHEN children > 0 OR babies > 0 THEN 'Con Hijos'
        ELSE 'Sin Hijos'
    END
ORDER BY tasa_cancelacion_pct DESC;
```

Se comparará la tasa de cancelación entre huéspedes con hijos y sin hijos. El segmento que presente el porcentaje más alto tendrá una mayor propensión a cancelar.

Ejercicio 18. Impacto de las solicitudes especiales

```python
%sql
SELECT
    total_of_special_requests,
    COUNT(*) AS total_reservas,
    SUM(is_canceled) AS reservas_canceladas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion_pct
FROM cursodataacademy.proyecto_hotel.hotel_bookings
GROUP BY total_of_special_requests
ORDER BY total_of_special_requests ASC;
```

Se analizará si la cantidad de solicitudes especiales está relacionada con una mayor o menor tasa de cancelación. Una tasa decreciente al aumentar las solicitudes podría sugerir un mayor compromiso del huésped.

Ejercicio 19. ¿Modificar la reserva es señal de compromiso?

```python
%sql
SELECT
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        WHEN booking_changes > 3 THEN 'Más de 3 cambios'
    END AS grupo_cambios,
    COUNT(*) AS total_reservas,
    SUM(is_canceled) AS reservas_canceladas,
    ROUND(AVG(is_canceled) * 100, 2) AS tasa_cancelacion_pct
FROM cursodataacademy.proyecto_hotel.hotel_bookings
GROUP BY
    CASE
        WHEN booking_changes = 0 THEN 'Sin Cambios'
        WHEN booking_changes BETWEEN 1 AND 3 THEN '1 a 3 cambios'
        WHEN booking_changes > 3 THEN 'Más de 3 cambios'
    END
ORDER BY tasa_cancelacion_pct ASC;
```

Ejercicio 20. Relación entre tipo de hotel, cancelaciones y ADR

```python
%sql
SELECT
    hotel,
    COUNT(*) AS cantidad_cancelaciones,
    ROUND(AVG(adr), 2) AS precio_promedio_cancelaciones
FROM cursodataacademy.proyecto_hotel.hotel_bookings
WHERE is_canceled = 1
GROUP BY hotel
ORDER BY precio_promedio_cancelaciones DESC;
```

La consulta compara la cantidad de cancelaciones y el ADR promedio de las reservas canceladas entre los tipos de hotel. El hotel con mayor ADR presenta un mayor precio promedio asociado a sus cancelaciones.
